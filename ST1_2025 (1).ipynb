```python
import numpy as np
import pandas as pd
import seaborn as sns

import matplotlib.pyplot as plt
%matplotlib inline

from sklearn.preprocessing import LabelEncoder
```

```python
carsData = pd.read_csv('cars.csv')
df = pd.DataFrame(carsData)

df.columns =['name','year','selling_price','km_driven','fuel','seller_type','transmission','owner']
# This dataframe has 4340 rows and 8 columns
df.shape
```

```python
# This is the total number of elements
print(df.shape[0] * df.shape[1])
```

```python
## This is shuffling from row 1 to 10
start, end =0,9

shuffled_cars_byRow= df.iloc[start:end].sample(frac=1,random_state=42)
print(shuffled_cars_byRow)
```

```python
## This is shuffling from index label 1 to 10
start, end =1,10

shuffled_cars_byIndex= df.iloc[start:end].sample(frac=1,random_state=42)
print(shuffled_cars_byIndex)

#The outputs differs because of the index starting points
```

```python
reseted_cars_byIndex=shuffled_cars_byIndex.reset_index(drop=True)
reseted_cars_byIndex
```

```python
reseted_cars_byRow=shuffled_cars_byRow.reset_index(drop=True)
reseted_cars_byRow
```

```python
missing_values=df.isnull().sum().sort_values(ascending=False)
percentage_missing_values= (missing_values/len(df)) * 100
pd.concat([missing_values, percentage_missing_values], axis = 1, keys= ['Missing values', '% Missing'])
```

```python
df[(df.selling_price.isnull()  == True)]
```

```python

df['selling_price']=df['selling_price'].fillna(df.groupby('name')['selling_price'].transform('mean'))
df.isnull().sum()
```

```python
df.drop(['name'],axis=1,inplace=True)
```

```python
print(f"Fuel {df['fuel'].unique()}\n")
print(f"Seller type: {df['seller_type'].unique()}\n")
print(f"Transmission: {df['transmission'].unique()}\n")
```

```python
label_encoder = LabelEncoder()

df['transmission'] = label_encoder.fit_transform(df['transmission'])
df['seller_type'] = label_encoder.fit_transform(df['seller_type'])
df['fuel'] = label_encoder.fit_transform(df['fuel'])

df.head(10)
```

```python
print(f"Fuel {df['fuel'].unique()}\n")
print(f"Seller type: {df['seller_type'].unique()}\n")
print(f"Transmission: {df['transmission'].unique()}\n")
```
