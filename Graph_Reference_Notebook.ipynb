# 📊 Data Visualisation Reference Notebook
## Seaborn vs Matplotlib — Side-by-Side Comparison
### Dataset: FIFA World Cup 2018 Match Results (`game_results.csv`)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Load dataset
df = pd.read_csv('game_results.csv')

# Quick prep
df['Man of the Match'] = df['Man of the Match'].map({'Yes': 1, 'No': 0})
df['PSO'] = df['PSO'].map({'Yes': 1, 'No': 0})

print('Dataset loaded. Shape:', df.shape)
df.head(3)
```

---
## 1. Count Plot
**When to use:** Count how many times each category appears in a column.

- ✅ **Seaborn** — `sns.countplot()` — built-in, one line
- ✅ **Matplotlib** — `value_counts()` + `plt.bar()`

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- SEABORN ---
sns.countplot(data=df, x='Round', order=df['Round'].value_counts().index,
              palette='Set2', edgecolor='black', ax=axes[0])
axes[0].set_title('Count Plot — SEABORN (sns.countplot)', fontsize=13, fontweight='bold', color='navy')
axes[0].set_xlabel('Round')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=30)

# --- MATPLOTLIB ---
counts = df['Round'].value_counts()
axes[1].bar(counts.index, counts.values, color='steelblue', edgecolor='black')
axes[1].set_title('Count Plot — MATPLOTLIB (plt.bar + value_counts)', fontsize=13, fontweight='bold', color='darkred')
axes[1].set_xlabel('Round')
axes[1].set_ylabel('Count')
axes[1].tick_params(axis='x', rotation=30)

plt.suptitle('1. COUNT PLOT', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

---
## 2. Bar Chart (Aggregated Values)
**When to use:** Compare a numeric metric across categories (e.g. average goals per round).

- ✅ **Seaborn** — `sns.barplot()` — includes confidence intervals automatically
- ✅ **Matplotlib** — `groupby().mean()` + `plt.bar()`

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- SEABORN ---
sns.barplot(data=df, x='Round', y='Goal Scored',
            order=df['Round'].value_counts().index,
            palette='muted', edgecolor='black', ax=axes[0])
axes[0].set_title('Bar Chart — SEABORN (sns.barplot)', fontsize=13, fontweight='bold', color='navy')
axes[0].set_xlabel('Round')
axes[0].set_ylabel('Avg Goals Scored')
axes[0].tick_params(axis='x', rotation=30)

# --- MATPLOTLIB ---
avg_goals = df.groupby('Round')['Goal Scored'].mean().reindex(df['Round'].value_counts().index)
axes[1].bar(avg_goals.index, avg_goals.values, color='coral', edgecolor='black')
axes[1].set_title('Bar Chart — MATPLOTLIB (groupby + plt.bar)', fontsize=13, fontweight='bold', color='darkred')
axes[1].set_xlabel('Round')
axes[1].set_ylabel('Avg Goals Scored')
axes[1].tick_params(axis='x', rotation=30)

plt.suptitle('2. BAR CHART (Aggregated)', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

---
## 3. Horizontal Bar Chart
**When to use:** Same as bar chart but easier to read when category labels are long.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
top_teams = df.groupby('Team')['Goal Scored'].sum().nlargest(10)

# --- SEABORN ---
sns.barplot(x=top_teams.values, y=top_teams.index,
            palette='Blues_d', edgecolor='black', ax=axes[0])
axes[0].set_title('Horizontal Bar — SEABORN (sns.barplot, orient h)', fontsize=12, fontweight='bold', color='navy')
axes[0].set_xlabel('Total Goals')

# --- MATPLOTLIB ---
axes[1].barh(top_teams.index, top_teams.values, color='teal', edgecolor='black')
axes[1].set_title('Horizontal Bar — MATPLOTLIB (plt.barh)', fontsize=12, fontweight='bold', color='darkred')
axes[1].set_xlabel('Total Goals')

plt.suptitle('3. HORIZONTAL BAR CHART', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

---
## 4. Pie Chart
**When to use:** Show proportions of a whole (best with ≤ 6 categories).

- ⚠️ **Seaborn** — does NOT have a native pie chart. Use **Matplotlib** only.
- ✅ **Matplotlib** — `plt.pie()`

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

round_counts = df['Round'].value_counts()

# --- SEABORN (workaround: countplot styled as proportion) ---
proportions = round_counts / round_counts.sum() * 100
sns.barplot(x=proportions.values, y=proportions.index,
            palette='pastel', edgecolor='black', ax=axes[0])
axes[0].set_title('Pie Alternative — SEABORN\n(no native pie; use horizontal barplot for %)', fontsize=11, fontweight='bold', color='navy')
axes[0].set_xlabel('Percentage (%)')
for i, v in enumerate(proportions.values):
    axes[0].text(v + 0.3, i, f'{v:.1f}%', va='center', fontsize=9)

# --- MATPLOTLIB ---
colours = plt.cm.Set3(np.linspace(0, 1, len(round_counts)))
axes[1].pie(round_counts.values, labels=round_counts.index,
            autopct='%1.1f%%', colors=colours,
            startangle=90, wedgeprops={'edgecolor': 'black'})
axes[1].set_title('Pie Chart — MATPLOTLIB (plt.pie)', fontsize=13, fontweight='bold', color='darkred')

plt.suptitle('4. PIE CHART', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

---
## 5. Histogram
**When to use:** Show the distribution/frequency of a single numeric variable.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- SEABORN ---
sns.histplot(data=df, x='Attempts', bins=15, kde=True,
             color='steelblue', edgecolor='black', ax=axes[0])
axes[0].set_title('Histogram — SEABORN (sns.histplot, kde=True)', fontsize=12, fontweight='bold', color='navy')
axes[0].set_xlabel('Attempts')

# --- MATPLOTLIB ---
axes[1].hist(df['Attempts'], bins=15, color='darkorange', edgecolor='black')
axes[1].set_title('Histogram — MATPLOTLIB (plt.hist)', fontsize=12, fontweight='bold', color='darkred')
axes[1].set_xlabel('Attempts')
axes[1].set_ylabel('Frequency')

plt.suptitle('5. HISTOGRAM', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

---
## 6. Box and Whiskers (Box Plot)
**When to use:** Show distribution, median, IQR, and outliers of a numeric variable across categories.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- SEABORN ---
sns.boxplot(data=df, x='Round', y='Goal Scored',
            order=df['Round'].value_counts().index,
            palette='Set2', width=0.5, ax=axes[0])
axes[0].set_title('Box Plot — SEABORN (sns.boxplot)', fontsize=13, fontweight='bold', color='navy')
axes[0].tick_params(axis='x', rotation=30)

# --- MATPLOTLIB ---
round_order = df['Round'].value_counts().index.tolist()
data_by_round = [df[df['Round'] == r]['Goal Scored'].values for r in round_order]
bp = axes[1].boxplot(data_by_round, labels=round_order, patch_artist=True)
colours_box = plt.cm.Set2(np.linspace(0, 1, len(round_order)))
for patch, colour in zip(bp['boxes'], colours_box):
    patch.set_facecolor(colour)
axes[1].set_title('Box Plot — MATPLOTLIB (plt.boxplot)', fontsize=13, fontweight='bold', color='darkred')
axes[1].set_xlabel('Round')
axes[1].set_ylabel('Goal Scored')
axes[1].tick_params(axis='x', rotation=30)

plt.suptitle('6. BOX AND WHISKERS PLOT', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

---
## 7. Violin Plot
**When to use:** Like a box plot but also shows the full distribution shape (density) on each side.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- SEABORN ---
sns.violinplot(data=df, x='Round', y='Attempts',
               order=df['Round'].value_counts().index,
               palette='muted', inner='quartile', ax=axes[0])
axes[0].set_title('Violin Plot — SEABORN (sns.violinplot)', fontsize=12, fontweight='bold', color='navy')
axes[0].tick_params(axis='x', rotation=30)

# --- MATPLOTLIB ---
round_order = df['Round'].value_counts().index.tolist()
data_by_round = [df[df['Round'] == r]['Attempts'].dropna().values for r in round_order]
vp = axes[1].violinplot(data_by_round, showmedians=True)
axes[1].set_xticks(range(1, len(round_order)+1))
axes[1].set_xticklabels(round_order, rotation=30)
axes[1].set_title('Violin Plot — MATPLOTLIB (plt.violinplot)', fontsize=12, fontweight='bold', color='darkred')
axes[1].set_ylabel('Attempts')

plt.suptitle('7. VIOLIN PLOT', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

---
## 8. Scatter Plot — Basic
**When to use:** Examine the relationship between two numeric variables.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- SEABORN ---
sns.scatterplot(data=df, x='Attempts', y='Goal Scored',
                color='steelblue', edgecolor='k', alpha=0.7, s=60, ax=axes[0])
axes[0].set_title('Scatter Plot — SEABORN (sns.scatterplot)', fontsize=12, fontweight='bold', color='navy')

# --- MATPLOTLIB ---
axes[1].scatter(df['Attempts'], df['Goal Scored'],
                c='darkorange', edgecolors='k', alpha=0.7, s=60)
axes[1].set_title('Scatter Plot — MATPLOTLIB (plt.scatter)', fontsize=12, fontweight='bold', color='darkred')
axes[1].set_xlabel('Attempts')
axes[1].set_ylabel('Goal Scored')

plt.suptitle('8. SCATTER PLOT — Basic', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

---
## 9. Scatter Plot — Coloured by Category (Hue)
**When to use:** Identify cluster or group patterns within a scatter plot.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
palette = {'Group Stage': '#e41a1c', 'Round of 16': '#377eb8',
           'Quarter-finals': '#4daf4a', 'Semi-finals': '#984ea3', 'Final': '#ff7f00'}

# --- SEABORN ---
sns.scatterplot(data=df, x='Attempts', y='Goal Scored',
                hue='Round', palette=palette, edgecolor='k', s=70, ax=axes[0])
axes[0].set_title('Scatter (Hue) — SEABORN', fontsize=12, fontweight='bold', color='navy')
axes[0].legend(title='Round', fontsize=8)

# --- MATPLOTLIB ---
for rnd, colour in palette.items():
    sub = df[df['Round'] == rnd]
    axes[1].scatter(sub['Attempts'], sub['Goal Scored'],
                    c=colour, edgecolors='k', s=70, alpha=0.8, label=rnd)
axes[1].set_title('Scatter (Hue) — MATPLOTLIB (loop per group)', fontsize=12, fontweight='bold', color='darkred')
axes[1].set_xlabel('Attempts')
axes[1].set_ylabel('Goal Scored')
axes[1].legend(title='Round', fontsize=8)

plt.suptitle('9. SCATTER PLOT — Coloured by Category', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

---
## 10. Scatter Plot — Size Encoding (Bubble Chart)
**When to use:** Encode a third numeric variable as the size of each point.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- SEABORN ---
sns.scatterplot(data=df, x='Attempts', y='Goal Scored',
                size='Passes', sizes=(20, 400),
                hue='Round', palette=palette, edgecolor='k', alpha=0.75, ax=axes[0])
axes[0].set_title('Bubble Chart — SEABORN (size= parameter)', fontsize=12, fontweight='bold', color='navy')
axes[0].legend(fontsize=7, ncol=2)

# --- MATPLOTLIB ---
sizes = (df['Passes'] - df['Passes'].min()) / (df['Passes'].max() - df['Passes'].min()) * 400 + 20
axes[1].scatter(df['Attempts'], df['Goal Scored'],
                s=sizes, c=df['Ball Possession %'],
                cmap='viridis', edgecolors='k', alpha=0.75)
axes[1].set_title('Bubble Chart — MATPLOTLIB (s= scaled size)', fontsize=12, fontweight='bold', color='darkred')
axes[1].set_xlabel('Attempts')
axes[1].set_ylabel('Goal Scored')

plt.suptitle('10. BUBBLE CHART (Scatter + Size Encoding)', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

---
## 11. Scatter Plot — Regression Line (lmplot / regplot)
**When to use:** Show the trend/linear relationship between two variables.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- SEABORN ---
sns.regplot(data=df, x='On-Target', y='Goal Scored',
            scatter_kws={'edgecolors': 'k', 'alpha': 0.6},
            line_kws={'color': 'red', 'linewidth': 2},
            color='steelblue', ax=axes[0])
axes[0].set_title('Regression Scatter — SEABORN (sns.regplot)', fontsize=12, fontweight='bold', color='navy')

# --- MATPLOTLIB ---
x = df['On-Target']
y = df['Goal Scored']
m, b = np.polyfit(x, y, 1)
axes[1].scatter(x, y, c='darkorange', edgecolors='k', alpha=0.6)
axes[1].plot(np.sort(x), m * np.sort(x) + b, color='red', linewidth=2)
axes[1].set_title('Regression Scatter — MATPLOTLIB (np.polyfit)', fontsize=12, fontweight='bold', color='darkred')
axes[1].set_xlabel('On-Target')
axes[1].set_ylabel('Goal Scored')

plt.suptitle('11. SCATTER PLOT — With Regression Line', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

## 12. Heatmap — Correlation Matrix
**When to use:** Visualise the pairwise correlation between numeric features.

- ✅ **Seaborn** — `sns.heatmap()` — native support
- ✅ **Matplotlib** — `plt.imshow()` + manual annotations

```python
numeric_cols = ['Goal Scored', 'Attempts', 'On-Target', 'Off-Target',
                'Corners', 'Fouls Committed', 'Yellow Card', 'Saves',
                'Pass Accuracy %', 'Ball Possession %']
corr = df[numeric_cols].corr()

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# --- SEABORN ---
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, linewidths=0.5, vmin=-1, vmax=1, ax=axes[0])
axes[0].set_title('Heatmap — SEABORN (sns.heatmap)', fontsize=13, fontweight='bold', color='navy')

# --- MATPLOTLIB ---
im = axes[1].imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
axes[1].set_xticks(range(len(numeric_cols)))
axes[1].set_yticks(range(len(numeric_cols)))
axes[1].set_xticklabels(numeric_cols, rotation=45, ha='right', fontsize=8)
axes[1].set_yticklabels(numeric_cols, fontsize=8)
for i in range(len(numeric_cols)):
    for j in range(len(numeric_cols)):
        axes[1].text(j, i, f'{corr.iloc[i, j]:.2f}', ha='center', va='center', fontsize=7)
plt.colorbar(im, ax=axes[1])
axes[1].set_title('Heatmap — MATPLOTLIB (plt.imshow)', fontsize=13, fontweight='bold', color='darkred')

plt.suptitle('12. HEATMAP — Correlation Matrix', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

## 14. KDE Plot (Kernel Density Estimate)
**When to use:** Smooth version of a histogram showing the probability density of a variable.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- SEABORN ---
sns.kdeplot(data=df, x='Attempts', fill=True, color='steelblue', alpha=0.5, ax=axes[0])
sns.kdeplot(data=df, x='On-Target', fill=True, color='darkorange', alpha=0.5, ax=axes[0])
axes[0].legend(['Attempts', 'On-Target'])
axes[0].set_title('KDE Plot — SEABORN (sns.kdeplot)', fontsize=12, fontweight='bold', color='navy')

# --- MATPLOTLIB (via numpy) ---
from scipy.stats import gaussian_kde
for col, colour in zip(['Attempts', 'On-Target'], ['steelblue', 'darkorange']):
    vals = df[col].dropna().values
    kde = gaussian_kde(vals)
    x_range = np.linspace(vals.min(), vals.max(), 200)
    axes[1].fill_between(x_range, kde(x_range), alpha=0.5, color=colour, label=col)
axes[1].legend()
axes[1].set_title('KDE Plot — MATPLOTLIB (scipy gaussian_kde)', fontsize=12, fontweight='bold', color='darkred')

plt.suptitle('14. KDE PLOT (Kernel Density Estimate)', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

## 15. Pair Plot (Scatter Matrix)
**When to use:** Explore all pairwise relationships between multiple numeric features at once.

- ✅ **Seaborn** — `sns.pairplot()` — one line
- ✅ **Matplotlib** — `pd.plotting.scatter_matrix()`

```python
pair_cols = ['Goal Scored', 'Attempts', 'On-Target', 'Saves']
pair_df = df[pair_cols + ['Round']].copy()

# --- SEABORN ---
print('--- SEABORN: sns.pairplot ---')
g = sns.pairplot(pair_df, hue='Round', diag_kind='kde',
                 plot_kws={'alpha': 0.6, 'edgecolor': 'k', 's': 40})
g.fig.suptitle('Pair Plot — SEABORN (sns.pairplot)', y=1.02, fontsize=13, fontweight='bold', color='navy')
plt.show()
```

```python
# --- MATPLOTLIB ---
print('--- MATPLOTLIB: pd.plotting.scatter_matrix ---')
from pandas.plotting import scatter_matrix
fig, _ = plt.subplots()
plt.close()
axes_sm = scatter_matrix(df[pair_cols], figsize=(10, 10), alpha=0.5,
                          diagonal='kde', color='darkorange', edgecolors='k')
plt.suptitle('Pair Plot — MATPLOTLIB (scatter_matrix)', fontsize=13, fontweight='bold', color='darkred')
plt.tight_layout()
plt.show()
```

## 16. Stacked Bar Chart
**When to use:** Compare part-to-whole relationships across multiple categories.

```python
card_data = df.groupby('Round')[['Yellow Card', 'Red']].sum().reindex(df['Round'].value_counts().index)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- SEABORN (stacked via two barplots) ---
x = range(len(card_data))
axes[0].bar(x, card_data['Yellow Card'], label='Yellow Card', color='#f4d03f', edgecolor='black')
axes[0].bar(x, card_data['Red'], bottom=card_data['Yellow Card'], label='Red Card', color='#e74c3c', edgecolor='black')
axes[0].set_xticks(list(x))
axes[0].set_xticklabels(card_data.index, rotation=30)
axes[0].legend()
axes[0].set_title('Stacked Bar — SEABORN style\n(Matplotlib axes, Seaborn palette)', fontsize=11, fontweight='bold', color='navy')

# --- MATPLOTLIB ---
card_data.plot(kind='bar', stacked=True, color=['#f4d03f', '#e74c3c'],
               edgecolor='black', ax=axes[1])
axes[1].set_title('Stacked Bar — MATPLOTLIB (df.plot stacked=True)', fontsize=11, fontweight='bold', color='darkred')
axes[1].tick_params(axis='x', rotation=30)
axes[1].set_xlabel('Round')

plt.suptitle('16. STACKED BAR CHART', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

## 17. Strip Plot
**When to use:** Like a scatter plot for one numeric and one categorical variable — shows every individual data point.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- SEABORN ---
sns.stripplot(data=df, x='Round', y='Goal Scored',
              order=df['Round'].value_counts().index,
              palette='Set1', jitter=True, edgecolor='k', linewidth=0.5, size=6, ax=axes[0])
axes[0].set_title('Strip Plot — SEABORN (sns.stripplot)', fontsize=12, fontweight='bold', color='navy')
axes[0].tick_params(axis='x', rotation=30)

# --- MATPLOTLIB ---
round_order = df['Round'].value_counts().index.tolist()
colours_strip = plt.cm.Set1(np.linspace(0, 1, len(round_order)))
for idx, (rnd, colour) in enumerate(zip(round_order, colours_strip)):
    sub = df[df['Round'] == rnd]['Goal Scored'].values
    jitter = np.random.uniform(-0.2, 0.2, size=len(sub))
    axes[1].scatter(np.full(len(sub), idx) + jitter, sub,
                    color=colour, edgecolors='k', linewidths=0.5, s=40, alpha=0.8)
axes[1].set_xticks(range(len(round_order)))
axes[1].set_xticklabels(round_order, rotation=30)
axes[1].set_title('Strip Plot — MATPLOTLIB (manual jitter)', fontsize=12, fontweight='bold', color='darkred')
axes[1].set_ylabel('Goal Scored')

plt.suptitle('17. STRIP PLOT', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```
